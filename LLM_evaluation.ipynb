```python
from rouge_score import rouge_scorer
from sacrebleu.metrics import BLEU
def evaluate_response(reference, generated):
    # TODO: tokenize, compute precision, recall, and f1 as described in the Approach
    bleu=BLEU(effective_order=True)
    b_score=bleu.corpus_score([generated],[[reference]]).score/100

    scorer=rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)
    scores=scorer.score(reference, generated)

    r2_score=scores['rouge2'].fmeasure

    product_rb=b_score*r2_score
    sum_rb=b_score+r2_score

    if sum_rb == 0:
        return 0.0

    f1_score=2*(product_rb/sum_rb)

    return f1_score




if __name__ == "__main__":
    reference = "your order will be delivered within three to five business days"
    generated = "your order should arrive in three to five working days"
    print(evaluate_response(reference, generated))
```
