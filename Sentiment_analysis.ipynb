```python
TEMPLATE_OPENERS = {
    "negative": "We are sorry about this and we want to fix it.",
    "positive": "Thank you so much for your kind words.",
}

def route_ticket(label, score):
    # TODO: apply the three confidence bands from the Approach above and
    # return {"decision": ..., "opener": ...}
    if score >= 0.9:
        return {"decision": "Trust", "opener": TEMPLATE_OPENERS[label]}
    elif score >=0.6 and score < 0.9:
        return {"decision": "Caution", "opener": TEMPLATE_OPENERS[label]}
    else:
        return {"decision": "Human_handoff", "opener": None}

if __name__ == "__main__":
    tickets = [
        {"message": "My package has not arrived even after two weeks.", "label": "negative", "score": 0.94},
        {"message": "Love the product, ordering again!", "label": "positive", "score": 0.65},
        {"message": "It's an average product I guess.", "label": "negative", "score": 0.52},
    ]
    for t in tickets:
        print(route_ticket(t["label"], t["score"]))
```
