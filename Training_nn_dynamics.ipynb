```python
data=[{"layer":"fc1", "weight_avg_abs":0.21, "grad_l2_norm":0.31 },
      {"layer":"fc2", "weight_avg_abs":0.19, "grad_l2_norm":0.28 },
      {"layer":"fc3", "weight_avg_abs":0.0003, "grad_l2_norm":0.004 },
      {"layer":"fc4", "weight_avg_abs":45.2, "grad_l2_norm":87.6 },
      {"layer":"bn1", "weight_avg_abs":0.15, "grad_l2_norm":0.19 },
      {"layer":"bn2", "weight_avg_abs":0.0001, "grad_l2_norm":0.0009 }]
```

```python
def diagnose_layers(layers):
    """
    layers: list of dicts, each with keys "layer", "weight_avg_abs", "grad_l2_norm"
    returns: list of dicts, each with keys "layer", "status", "recommendation"
    """
    results = []
    for record in layers:
        if record["weight_avg_abs"] > 10 or record["grad_l2_norm"] > 10:
            results.append({"layer": record["layer"], "status": "EXPLODING", "recommendation": "gradient clipping, batch normalization, smaller learning rate, or better weight initialization"})
        elif record["weight_avg_abs"] <0.01 or record["grad_l2_norm"] < 0.001:
            results.append({"layer": record["layer"], "status": "VANISHING", "recommendation": "proper weight initialization (Xavier/He), batch normalization, or ReLU/Leaky ReLU activation -- NOT gradient clipping, since it does not fix vanishing gradients"})
        else:
            results.append({"layer": record["layer"], "status": "HEALTHY", "recommendation": "no action required"})
    return results

if __name__ == "__main__":
    for entry in diagnose_layers(data):
        print(f'{entry["layer"]} -> {entry["status"]} -> {entry["recommendation"]}')
```
