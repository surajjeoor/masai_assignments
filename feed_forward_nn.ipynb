```python
import numpy as np

def relu(z):
  return np.maximum(0, z)

def sigmoid(z):
  return 1/(1+np.exp(-z))

def forward(W1,W2,b1,b2,X):
  Z1 = np.dot(X,W1)+b1
  A1 = relu(Z1)
  Z2 = np.dot(A1,W2)+b2 # Changed from np.dot(W2,A1) to np.dot(A1,W2)
  A2 = sigmoid(Z2)
  return A2 # Added return statement as per previous turn's fix

def binary_crossentropy(y_true,y_pred):
  return -np.mean(y_true*np.log(y_pred)+(1-y_true)*np.log(1-y_pred))


if __name__=="__main__":
  X=np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
  y=np.array([[0],  [1], [1], [0]])
  W1=np.array([[0.5, -0.5], [0.5, 0.5]])
  b1=np.array([0.0, 0.0])
  W2=np.array([[1.0], [1.0]])
  b2=np.array([0.0])
  yhat=forward(W1,W2,b1,b2,X)
  loss=binary_crossentropy(y,yhat)
  print(loss)
```
