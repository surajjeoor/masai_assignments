```python
# Reference data: SHAP contributions for one applicant (percentage points)
BASELINE = 40
CONTRIBUTIONS = {
    "income": 18,
    "credit_history": 12,
    "existing_debt": -8,
    "loan_amount": -4,
    "other": -3,
}


def explain(baseline, contributions):
    # TODO: add every contribution to the baseline
    final = baseline
    raised = []
    lowered = []
    # TODO: sort factors by absolute contribution, largest first
    for factor, value in contributions.items():
        final += value
        if value > 0:
            raised.append(f"{factor} (+{value})")
        else:
            lowered.append(f"{factor} ({value})")


    # TODO: build "name (+value)" strings for raising and lowering factors

    return final, raised, lowered


if __name__ == "__main__":
    final, raised, lowered = explain(BASELINE, CONTRIBUTIONS)
    print(f"Final approval probability: {final}")
    print("Raised the score: " + ", ".join(raised))
    print("Lowered the score: " + ", ".join(lowered))
```

```python
# Sample input: (group, qualified, recommended)
RECORDS = [
    ("A", 1, 1), ("B", 1, 1), ("A", 0, 0), ("B", 0, 0), ("A", 1, 1),
    ("B", 1, 0), ("A", 0, 1), ("B", 0, 0), ("A", 1, 1), ("B", 1, 1),
    ("A", 0, 0), ("B", 0, 0), ("A", 1, 0), ("B", 1, 1), ("A", 1, 1),
    ("B", 1, 0), ("A", 0, 0), ("B", 0, 0), ("A", 1, 1), ("B", 0, 0),
]


def group_rates(records, group):
    # TODO: count TP, FN, FP and TN for this group
    tp_count=0
    fp_count=0
    tn_count=0
    fn_count=0
    selection_rate=0.0
    TPR=0.0
    FPR=0.0
    for record in records:
        if record[0] == group:
            if record[1] == 1:
                if record[2] == 1:
                    tp_count += 1
                else:
                    fn_count += 1
            else:
                if record[2] == 1:
                    fp_count += 1
                else:
                    tn_count += 1

        else:
            continue
    # TODO: return the selection rate, TPR and FPR
    selection_rate = (tp_count + fn_count) / (tp_count + fp_count + tn_count + fn_count)
    TPR = tp_count / (tp_count + fn_count)
    FPR = fp_count / (fp_count + tn_count)
    return selection_rate, TPR, FPR


def audit(records):
    # TODO: print each group's rates
    for group in ["A", "B"]:
        selection_rate, TPR, FPR = group_rates(records, group)
        print(f"Group {group}:")
        print(f"  Selection rate: {selection_rate} \t TPR: {TPR} \t FPR: {FPR}")
    # TODO: print the parity difference, the ratio, the max gap and the verdict
    selection_rate_A, TPR_A, FPR_A = group_rates(records, "A")
    selection_rate_B, TPR_B, FPR_B = group_rates(records, "B")
    print(f"Parity difference: {TPR_A - TPR_B}")
    print(f"Ratio: {TPR_B / TPR_A}")
    print(f"Max gap: {max(abs(TPR_A - TPR_B), abs(FPR_A - FPR_B))}")
    #check the fourth fifths test
    print("Passes fourth fifths rule:",abs(TPR_B/TPR_A)>=0.80)
    pass


if __name__ == "__main__":
    audit(RECORDS)
```
